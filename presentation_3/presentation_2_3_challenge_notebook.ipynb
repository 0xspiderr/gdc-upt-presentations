```python
# @title
from __future__ import annotations


def validate_tool_enum(Tool):
    def run():
        return (
            tuple(member.name for member in Tool),
            tuple(member.value for member in Tool),
            len(Tool),
            Tool.WATER.name,
            Tool.SHOVEL.name,
            Tool.WATER.value,
            Tool.SHOVEL.value,
            Tool(0) == Tool.WATER,
            Tool(1) == Tool.SHOVEL,
            Tool.WATER < Tool.SHOVEL,
        )

    assert run() == (
        ("WATER", "SHOVEL"),
        (0, 1),
        2,
        "WATER",
        "SHOVEL",
        0,
        1,
        True,
        True,
        True,
    )
    return True


def validate_state_enum(State):
    def run():
        return (
            tuple(member.name for member in State),
            tuple(member.value for member in State),
            len(State),
            State.IDLE.name,
            State.WALKING.name,
            State.RUNNING.name,
            State.USING_TOOL.name,
            State(0) == State.IDLE,
            State(1) == State.WALKING,
            State(3) == State.USING_TOOL,
        )

    assert run() == (
        ("IDLE", "WALKING", "RUNNING", "USING_TOOL"),
        (0, 1, 2, 3),
        4,
        "IDLE",
        "WALKING",
        "RUNNING",
        "USING_TOOL",
        True,
        True,
        True,
    )
    return True


def validate_current_speed(current_speed):
    def run():
        cases = (
            (False, 120, 220),
            (True, 120, 220),
            (False, 90, 150),
            (True, 90, 150),
            (False, 1, 2),
            (True, 1, 2),
            (False, 300, 500),
            (True, 300, 500),
            (False, 0, 999),
            (True, 0, 999),
        )
        return tuple(current_speed(*case) for case in cases)

    assert run() == (120, 220, 90, 150, 1, 2, 300, 500, 0, 999)
    return True


def validate_player_move(player_move):
    def run():
        cases = (
            (0.0, 0.0, 1.0),
            (4.0, 8.0, 0.5),
            (10.0, -2.0, 2.0),
            (-5.0, 5.0, 1.0),
            (100.0, 12.5, 0.0),
            (1.5, 2.5, 3.0),
            (7.0, 1.0, 4.0),
            (9.0, -3.0, 3.0),
            (2.0, 2.0, 2.0),
            (-10.0, -5.0, 0.5),
        )
        return tuple(player_move(*case) for case in cases)

    assert run() == (0.0, 8.0, 6.0, 0.0, 100.0, 9.0, 11.0, 0.0, 6.0, -12.5)
    return True


def validate_tool_animation_for_player_state(tool_animation_for_player_state):
    def run():
        cases = (
            ("idle", None),
            ("run", None),
            ("idle", "water"),
            ("run", "shovel"),
            ("idle", "water"),
            ("run", "water"),
            ("idle", "shovel"),
            ("run", "shovel"),
            ("idle", None),
            ("run", None),
        )
        return tuple(tool_animation_for_player_state(*case) for case in cases)

    assert run() == ("idle", "run", "water", "shovel", "water", "water", "shovel", "shovel", "idle", "run")
    return True


def validate_player_animate(player_animate, AnimationMock):
    def run():
        body_player = AnimationMock()
        tool_player = AnimationMock()
        result = player_animate("run", body_player, tool_player)
        return (
            player_animate.__annotations__["body_player"],
            player_animate.__annotations__["tool_player"],
            result,
            tuple(body_player.played),
            tuple(tool_player.played),
        )

    assert run() == (AnimationMock, AnimationMock, None, ("run",), ("run",))
    return True


def validate_choose_state(choose_state, State):
    def run():
        cases = (
            (0, False, False),
            (3, False, False),
            (3, True, False),
            (3, True, True),
            (1, False, False),
            (1, True, False),
            (0, True, False),
            (0, False, True),
            (9, False, True),
            (9, True, True),
        )
        return tuple(choose_state(*case) for case in cases)

    assert run() == (
        State.IDLE,
        State.WALKING,
        State.RUNNING,
        State.USING_TOOL,
        State.WALKING,
        State.RUNNING,
        State.IDLE,
        State.USING_TOOL,
        State.USING_TOOL,
        State.USING_TOOL,
    )
    return True


def validate_apply_requested_state(apply_requested_state, State):
    def run():
        cases = (
            (State.IDLE, State.RUNNING, False),
            (State.IDLE, State.RUNNING, True),
            (State.WALKING, State.IDLE, False),
            (State.WALKING, State.IDLE, True),
            (State.RUNNING, State.USING_TOOL, False),
            (State.RUNNING, State.USING_TOOL, True),
            (State.USING_TOOL, State.IDLE, False),
            (State.USING_TOOL, State.IDLE, True),
            (State.IDLE, State.WALKING, False),
            (State.IDLE, State.WALKING, True),
        )
        return tuple(apply_requested_state(*case) for case in cases)

    assert run() == (
        State.RUNNING,
        State.IDLE,
        State.IDLE,
        State.WALKING,
        State.USING_TOOL,
        State.RUNNING,
        State.IDLE,
        State.USING_TOOL,
        State.WALKING,
        State.IDLE,
    )
    return True


def validate_switch_tool(switch_tool):
    def run():
        cases = (
            (0, 1, 2),
            (1, 1, 2),
            (0, -1, 2),
            (1, -1, 2),
            (0, 1, 4),
            (3, 1, 4),
            (0, -1, 4),
            (3, -1, 4),
            (2, 1, 5),
            (2, -1, 5),
        )
        return tuple(switch_tool(*case) for case in cases)

    assert run() == (1, 0, 1, 0, 1, 0, 3, 2, 3, 1)
    return True


def validate_describe_state(describe_state, State):
    def run():
        cases = (
            (State.IDLE, State.IDLE),
            (State.IDLE, State.RUNNING),
            (State.WALKING, State.USING_TOOL),
            (State.RUNNING, State.USING_TOOL),
            (State.WALKING, State.RUNNING),
            (State.RUNNING, State.WALKING),
            (State.USING_TOOL, State.IDLE),
            (State.IDLE, State.USING_TOOL),
            (State.WALKING, State.IDLE),
            (State.USING_TOOL, State.RUNNING),
        )
        return tuple(describe_state(*case) for case in cases)

    assert run() == (
        State.IDLE,
        State.IDLE,
        State.WALKING,
        State.RUNNING,
        State.RUNNING,
        State.WALKING,
        State.IDLE,
        State.USING_TOOL,
        State.IDLE,
        State.RUNNING,
    )
    return True


def validate_play_tool_animation(play_tool_animation, AnimationMock):
    def run():
        body_player = AnimationMock()
        tool_player = AnimationMock()
        result = play_tool_animation("WATER", body_player, tool_player)
        return (
            play_tool_animation.__annotations__["body_player"],
            play_tool_animation.__annotations__["tool_player"],
            result,
            tuple(body_player.played),
            tuple(tool_player.played),
        )

    assert run() == (AnimationMock, AnimationMock, None, ("water",), ("water",))
    return True


def validate_facing_controller(FacingController, SpriteMock):
    def run():
        body = SpriteMock()
        tool = SpriteMock()
        controller = FacingController(body, tool)
        annotations = (
            FacingController.__init__.__annotations__["body_sprite"],
            FacingController.__init__.__annotations__["tool_sprite"],
        )
        controller.set_facing_right(True)
        first = (body.flip_h, tool.flip_h)
        controller.set_facing_right(False)
        second = (body.flip_h, tool.flip_h)
        return annotations, first, second

    assert run() == (
        (
            SpriteMock,
            SpriteMock,
        ),
        (False, False),
        (True, True),
    )
    return True


def validate_tool_use_controller(ToolUseController):
    def run():
        controller = ToolUseController()
        states = [controller.can_move]
        for _ in range(5):
            controller.start_tool_use()
            states.append(controller.can_move)
            controller.on_animation_finished()
            states.append(controller.can_move)
        return tuple(states)

    assert run() == (
        True,
        False,
        True,
        False,
        True,
        False,
        True,
        False,
        True,
        False,
        True,
    )
    return True


def validate_signal(Signal, Tool):
    def run():
        class LevelMock:
            def __init__(self):
                self.received = []

            def on_tool_used(self, tool):
                self.received.append(tool.name)

        signal = Signal()
        level = LevelMock()
        signal.connect(level.on_tool_used)
        for tool in (
            Tool.WATER,
            Tool.SHOVEL,
            Tool.WATER,
            Tool.SHOVEL,
            Tool.WATER,
            Tool.SHOVEL,
            Tool.WATER,
            Tool.SHOVEL,
            Tool.WATER,
            Tool.SHOVEL,
        ):
            signal.emit(tool)
        return tuple(level.received)

    assert run() == (
        "WATER",
        "SHOVEL",
        "WATER",
        "SHOVEL",
        "WATER",
        "SHOVEL",
        "WATER",
        "SHOVEL",
        "WATER",
        "SHOVEL",
    )
    return True
```
